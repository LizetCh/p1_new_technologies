# 02 Ingestion Test

This notebook is used to test the manual ingestion logic before converting it into an automated Python script.

For each table, the structure should be:

```text
1. Read the CSV file
2. Add metadata columns
3. Load the data into the raw PostgreSQL table
```

## 1. Connect to PostgreSQL

This section imports the required libraries and creates the database connection.

````python
import pandas as pd
from sqlalchemy import create_engine
````

````python
engine = create_engine(
    "postgresql+psycopg2://postgres:1603@localhost:5432/hospital_management"
)
````

## 2. Load patients CSV into the raw table

This cell reads the patients CSV file, adds the `source_file` metadata column, and loads the data into the PostgreSQL table `raw.patients`.

The purpose of this step is to test the ingestion process manually before moving the logic into the automated ingestion script.

````python
df = pd.read_csv("../data/csv_files/1.ingestion/manual-ingestion/patients00.csv")
df.head()
````

````python
df["source_file"] = "patients00.csv"
##
````

````python
df.head()
````

````python
df.to_sql(
    "patients",
      engine, 
      schema="raw", 
      if_exists="append", 
      index=False)
````
